```python
from datasets import load_dataset
import pandas as pd
import os

PROJECT_DIR = "multilingual_llm_kikuyu_luo_kamba_swahili"
os.makedirs(PROJECT_DIR, exist_ok=True)

languages = {
    "swh_Latn": "swahili",
    "luo_Latn": "luo",
    "kik_Latn": "kikuyu",
    "kam_Latn": "kamba",
}

# English reference with metadata
eng = load_dataset("openlanguagedata/flores_plus", "eng_Latn")

for lang_code, lang_name in languages.items():
    print(f"Processing {lang_name}...")
    ds = load_dataset("openlanguagedata/flores_plus", lang_code)

    for split in ["dev", "devtest"]:
        eng_df = pd.DataFrame({
            "id": eng[split]["id"],
            "english": eng[split]["text"],
            "domain": eng[split]["domain"],
            "topic": eng[split]["topic"],
        })
        lang_df = pd.DataFrame({
            "id": ds[split]["id"],
            lang_name: ds[split]["text"],
        })

        # Merge on id — guaranteed alignment
        df = eng_df.merge(lang_df, on="id", how="inner").sort_values("id")
        df.to_csv(f"{PROJECT_DIR}/{lang_name}_{split}.csv", index=False)
        print(f"  ✓ {split}: {len(df)} rows, "
              f"{df[lang_name].notna().all()} aligned")
```

```python
import requests

def check_opus(lang):
    url = f"https://opus.nlpl.eu/opusapi/?source={lang}&target=eng&preprocessing=moses&version=latest"
    for corpus in requests.get(url).json().get("corpus", []):
        print(f"  {corpus['name']}: {corpus.get('alignment_pairs', '?')} pairs")

for lang in ["kik", "luo", "kam", "swh"]:
    print(f"\n=== {lang} ==="); check_opus(lang)
```

```python
def clean_pairs(df, drop_q=0.10):
    df = df.dropna(subset=["src", "tgt"])
    df = df[df["src"].str.split().str.len().between(3, 250)]
    df = df[df["laser_score"] >= df["laser_score"].quantile(drop_q)]  # drop noisiest 10%
    return df.drop_duplicates(subset=["src", "tgt"])
```

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
import torch

tok = AutoTokenizer.from_pretrained("facebook/nllb-200-distilled-600M")

device = "cuda" if torch.cuda.is_available() else "cpu"
model = AutoModelForSeq2SeqLM.from_pretrained("facebook/nllb-200-distilled-600M").to(device).eval()

def nllb_translate(texts, src="eng_Latn", tgt="kik_Latn"):
    tok.src_lang = src
    batch = tok(texts, return_tensors="pt", padding=True, truncation=True, max_length=256).to(device)
    with torch.no_grad():
        out = model.generate(**batch, forced_bos_token_id=tok.lang_code_to_id[tgt],
                             max_new_tokens=256, num_beams=4)
    return tok.batch_decode(out, skip_special_tokens=True)
```

```python
import requests

def check_opus(src, tgt="eng"):
    url = f"https://opus.nlpl.eu/opusapi/?source={src}&target={tgt}&preprocessing=moses&version=latest"
    try:
        r = requests.get(url, timeout=30)
        data = r.json()
        if "corpus" not in data:
            print(f"  [{src}] no 'corpus' key. Raw response: {str(data)[:300]}")
            return
        for c in data["corpus"]:
            print(f"  {c['name']}: {c.get('alignment_pairs', '?')} pairs")
    except Exception as e:
        print(f"  [{src}] FAILED: {e}")

for code in ["sw", "swh", "kik", "luo", "kam"]:  # OPUS uses 2-letter codes when they exist
    print(f"=== {code} ==="); check_opus(code)
```

```python
from huggingface_hub import list_repo_files, hf_hub_download
from datasets import load_dataset
import pandas as pd

files = list_repo_files("allenai/nllb", repo_type="dataset", revision="refs/convert/parquet")
pair_dirs = {f.split("/")[0] for f in files if f.endswith(".parquet")}

my_langs = ["eng_Latn", "swh_Latn", "luo_Latn", "kik_Latn", "kam_Latn"]
wanted = []
for i, a in enumerate(my_langs):
    for b in my_langs[i+1:]:
        for p in (f"{a}-{b}", f"{b}-{a}"):
            if p in pair_dirs:
                wanted.append(p)
print(f"Found {len(wanted)} relevant pairs:", wanted)

MAX_SHARDS = 5  # eng-swh has ~57; don't download them all
for pair in wanted:
    shards = sorted(f for f in files if f.startswith(pair + "/") and f.endswith(".parquet"))[:MAX_SHARDS]
    frames = []
    for sh in shards:
        path = hf_hub_download("allenai/nllb", sh, repo_type="dataset", revision="refs/convert/parquet")
        ds = load_dataset("parquet", data_files=path, split="train")
        a, b = pair.split("-")
        frames.append(pd.DataFrame({
            "src_lang": a, "tgt_lang": b,
            "src": [t[a] for t in ds["translation"]],
            "tgt": [t[b] for t in ds["translation"]],
            "laser_score": ds["laser_score"],
        }))
    df = pd.concat(frames).reset_index(drop=True)
    df.to_csv(f"{PROJECT_DIR}/nllb_{pair}_train.csv", index=False)
    print(f"{pair}: {len(df):,} pairs saved")
```

```python
!pip install sacrebleu
import pandas as pd, torch, re, sacrebleu, os

# English pool: English side of your eng-swh gold (already clean, FLORES-domain)
pool_df = pd.read_csv(f"{PROJECT_DIR}/nllb_eng_Latn-swh_Latn_train.csv")
assert pool_df["src_lang"].iloc[0] == "eng_Latn"

norm = lambda s: re.sub(r"\s+", " ", str(s).strip().lower())
eval_sents = set()
for lang in ["swahili", "luo", "kikuyu", "kamba"]:
    for split in ["dev", "devtest"]:
        d = pd.read_csv(f"{PROJECT_DIR}/{lang}_{split}.csv")
        eval_sents.update(d["english"].map(norm)); eval_sents.update(d[lang].map(norm))

pool = (pool_df["src"].dropna().drop_duplicates()
        .loc[lambda s: s.str.split().str.len().between(5, 60)]
        .loc[lambda s: ~s.map(norm).isin(eval_sents)]
        .sample(30000, random_state=42).reset_index(drop=True))
print(f"English pool: {len(pool):,}")

from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
tok = AutoTokenizer.from_pretrained("facebook/nllb-200-distilled-600M")
model = AutoModelForSeq2SeqLM.from_pretrained(
    "facebook/nllb-200-distilled-600M").half().cuda().eval()

def translate(texts, src, tgt, bs=64, max_new=128):
    tok.src_lang = src
    forced = tok.convert_tokens_to_ids(tgt)
    order = sorted(range(len(texts)), key=lambda i: len(str(texts[i])))
    out = [""] * len(texts)
    for s in range(0, len(order), bs):
        idx = order[s:s+bs]
        enc = tok([str(texts[i]) for i in idx], return_tensors="pt", padding=True,
                  truncation=True, max_length=256).to("cuda")
        with torch.no_grad():
            gen = model.generate(**enc, forced_bos_token_id=forced,
                                 max_new_tokens=max_new, num_beams=1)
        for i, g in zip(idx, tok.batch_decode(gen, skip_special_tokens=True)):
            out[i] = g
    return out

# Translate with checkpoints — survives Colab disconnects, just rerun the cell
SILVER_F = f"{PROJECT_DIR}/silver_eng_kam_raw.csv"
done = len(pd.read_csv(SILVER_F)) if os.path.exists(SILVER_F) else 0
while done < len(pool):
    batch = pool.iloc[done:done+5000].tolist()
    pd.DataFrame({"english": batch, "kamba": translate(batch, "eng_Latn", "kam_Latn")}
                 ).to_csv(SILVER_F, mode="a", header=not os.path.exists(SILVER_F), index=False)
    done += len(batch); print(f"{done}/{len(pool)}")

# Round-trip filter: Kamba -> English, keep high-agreement pairs
silver = pd.read_csv(SILVER_F).dropna()
back = translate(silver["kamba"].tolist(), "kam_Latn", "eng_Latn")
silver["rt_chrf"] = [sacrebleu.sentence_chrf(h, [r]).score
                     for h, r in zip(back, silver["english"])]
ratio = silver["kamba"].str.split().str.len() / silver["english"].str.split().str.len()
final = silver[(silver["rt_chrf"] >= 40) & ratio.between(0.4, 2.5)].reset_index(drop=True)
final.to_csv(f"{PROJECT_DIR}/silver_eng_kam_filtered.csv", index=False)
print(f"Kept {len(final):,}/{len(silver):,} ({100*len(final)/len(silver):.0f}%) — "
      f"if survival < 60%, rerun with threshold 35")
```

```python
results = []
for lang, code in [("swahili","swh_Latn"),("luo","luo_Latn"),
                   ("kikuyu","kik_Latn"),("kamba","kam_Latn")]:
    d = pd.read_csv(f"{PROJECT_DIR}/{lang}_devtest.csv")
    for name, texts, ref_col, s, t in [
        (f"eng->{lang}", d["english"].tolist(), lang, "eng_Latn", code),
        (f"{lang}->eng", d[lang].tolist(), "english", code, "eng_Latn"),
    ]:
        hyps = translate(texts, s, t, bs=48, max_new=200)
        pd.DataFrame({"hyp": hyps, "ref": d[ref_col]}).to_csv(
            f"{PROJECT_DIR}/baseline_{name.replace('->','_')}.csv", index=False)
        chrf = sacrebleu.corpus_chrf(hyps, [d[ref_col].tolist()], word_order=2)
        results.append({"direction": name, "chrF++": round(chrf.score, 2)})
        print(results[-1])
pd.DataFrame(results).to_csv(f"{PROJECT_DIR}/baseline_nllb600m.csv", index=False)
```
