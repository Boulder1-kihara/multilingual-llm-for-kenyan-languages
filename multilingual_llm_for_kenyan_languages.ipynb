```python
from datasets import load_dataset
import pandas as pd
import os

PROJECT_DIR = "multilingual_llm_kikuyu_luo_kamba_swahili"
os.makedirs(PROJECT_DIR, exist_ok=True)

languages = {
    "swh_Latn": "swahili",
    "luo_Latn": "luo",
    "kik_Latn": "kikuyu",
    "kam_Latn": "kamba",
}

# English reference with metadata
eng = load_dataset("openlanguagedata/flores_plus", "eng_Latn")

for lang_code, lang_name in languages.items():
    print(f"Processing {lang_name}...")
    ds = load_dataset("openlanguagedata/flores_plus", lang_code)

    for split in ["dev", "devtest"]:
        eng_df = pd.DataFrame({
            "id": eng[split]["id"],
            "english": eng[split]["text"],
            "domain": eng[split]["domain"],
            "topic": eng[split]["topic"],
        })
        lang_df = pd.DataFrame({
            "id": ds[split]["id"],
            lang_name: ds[split]["text"],
        })

        # Merge on id — guaranteed alignment
        df = eng_df.merge(lang_df, on="id", how="inner").sort_values("id")
        df.to_csv(f"{PROJECT_DIR}/{lang_name}_{split}.csv", index=False)
        print(f"  ✓ {split}: {len(df)} rows, "
              f"{df[lang_name].notna().all()} aligned")
```

```python
import requests

def check_opus(lang):
    url = f"https://opus.nlpl.eu/opusapi/?source={lang}&target=eng&preprocessing=moses&version=latest"
    for corpus in requests.get(url).json().get("corpus", []):
        print(f"  {corpus['name']}: {corpus.get('alignment_pairs', '?')} pairs")

for lang in ["kik", "luo", "kam", "swh"]:
    print(f"\n=== {lang} ==="); check_opus(lang)
```

```python
def clean_pairs(df, drop_q=0.10):
    df = df.dropna(subset=["src", "tgt"])
    df = df[df["src"].str.split().str.len().between(3, 250)]
    df = df[df["laser_score"] >= df["laser_score"].quantile(drop_q)]  # drop noisiest 10%
    return df.drop_duplicates(subset=["src", "tgt"])
```

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
import torch

tok = AutoTokenizer.from_pretrained("facebook/nllb-200-distilled-600M")

device = "cuda" if torch.cuda.is_available() else "cpu"
model = AutoModelForSeq2SeqLM.from_pretrained("facebook/nllb-200-distilled-600M").to(device).eval()

def nllb_translate(texts, src="eng_Latn", tgt="kik_Latn"):
    tok.src_lang = src
    batch = tok(texts, return_tensors="pt", padding=True, truncation=True, max_length=256).to(device)
    with torch.no_grad():
        out = model.generate(**batch, forced_bos_token_id=tok.lang_code_to_id[tgt],
                             max_new_tokens=256, num_beams=4)
    return tok.batch_decode(out, skip_special_tokens=True)
```
